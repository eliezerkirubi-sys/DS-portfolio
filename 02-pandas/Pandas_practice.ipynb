```python
#conditional filtering
anime_data_extracted = anime_data_extracted[anime_data_extracted['Score'] != 'unknown']
anime_data_extracted.head()
```

```python
# pulling the required coloumns including 'source' and 'members
cols_to_use_extended = ['MAL_ID', 'Name', 'Score', 'Genres', 'Type', 'Aired', 'Source', 'Members']
anime_data_extracted = anime_data[cols_to_use_extended]

# extracting data where the source is manga and must be greater than 100000
filtered_anime = anime_data_extracted[
    (anime_data_extracted['Source'] == 'Manga') &
    (anime_data_extracted['Members'] > 100000)
]

# viewing the results
display(filtered_anime.head())
```

```python
# sorting of values
anime_data_extracted = anime_data_extracted.sort_values(by='Score', ascending=False)

# Load the other two datasets so they are defined in memory
anime_list = pd.read_csv('MyAnimeList-Database-master/data/animelist.csv')
anime_synop = pd.read_csv('MyAnimeList-Database-master/data/anime_with_synopsis.csv')

# trying to merge three data sets
print("Anime Data Length:", len(anime_data))
display(anime_data.head(3))

print("Anime List Length:", len(anime_list))
display(anime_list.head(3))

print("Anime Synopsis Length:", len(anime_synop))
display(anime_synop.head(3))
```

```python
# data merging (inner joint)
pd.merge(anime_data_extracted, anime_synop, on='MAL_ID')

# preparing and using the left outer join two data sets
#data 1
data1 = {
    'id': ['100', '101', '102', '103', '104', '106', '108', '110', '111', '113'],
    'city': ['Tokyo', 'Osaka', 'Kyoto', 'Hokkaido', 'Tokyo', 'Tokyo', 'Osaka', 'Kyoto', 'Hokkaido', 'Tokyo'],
    'birth_year': [1990, 1989, 1992, 1997, 1982, 1991, 1988, 1990, 1988, 1990],
    'name': ['Hiroshi', 'Akiko', 'Yuki', 'Satoru', 'Steeve', 'Mituru', 'Aoi', 'Tarou', 'Suguru', 'Mitsuo']
}
df1 = DataFrame(data1)
df1
```
