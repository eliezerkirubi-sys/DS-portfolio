```python
# Import
import requests, zipfile
import io

# Fetching automobile price data
url = 'http://archive.ics.uci.edu/ml/machine-learning-databases/autos/imports-85.data'
res = requests.get(url).content

# Loading  the fetched data into a DataFrame object
auto = pd.read_csv(io.StringIO(res.decode('utf-8')), header=None)

# Setting  labels for the data columns
auto.columns =['symboling', 'normalized-losses', 'make', 'fuel-type', 'aspiration', 'num-of-doors',
                            'body-style', 'drive-wheels', 'engine-location', 'wheel-base', 'length', 'width', 'height',
                            'curb-weight', 'engine-type', 'num-of-cylinders', 'engine-size', 'fuel-system', 'bore',
                            'stroke', 'compression-ratio', 'horsepower', 'peak-rpm', 'city-mpg', 'highway-mpg', 'price']
```

```python
# checking the type of data rows and coloums respectively
print('Shape of the car data:{}'.format (auto.shape))

auto.head() # displaying the first five rows
```

```python
# Processing the data
# Using horsepower , width and height as the explanatory variables

# counting the number of ? in the dataset
auto=auto[['price','horsepower','width','height']]
auto.isin(['?']).sum()


# Replacing "?" with NaN and drops down with NaN
auto= auto.replace('?',np.nan).dropna()
print('shape of the car data:{}'.format(auto.shape)) # the number of rows have reduced

#checking the data types before type conversion
print('Check data types\n{}\n'.format(auto.dtypes))
```

```python
# converting the object data types to numeric values
auto= auto.assign(price=pd.to_numeric(auto.price))
auto= auto.assign(horsepower=pd.to_numeric(auto.horsepower))
auto= auto.assign(width=pd.to_numeric(auto.width))

#correlation check
auto.corr()
```

```python
# Decision Trees (Mushroom Data Set) Determining if a group of mushrooms are poisonous or not based on various attributes

#Retrieving the data
url = 'http://archive.ics.uci.edu/ml/machine-learning-databases/mushroom/agaricus-lepiota.data'
res = requests.get(url).content

#Loading the retrieved data into a DataFrame object
mushroom= pd.read_csv(io.StringIO(res.decode('utf-8')), header=None)

#setting coloumns labels
mushroom.columns = [  'classes', 'cap_shape', 'cap_surface', 'cap_color', 'bruises', 'odor',
                       'gill_attachment','gill_spacing','gill_size','gill_color','stalk_shape',
                        'stalk_root','stalk_surface_above_ring','stalk_surface_below_ring',
                        'stalk_color_above_ring','stalk_color_below_ring','veil_type','veil_color',
                        'ring_number','ring_type','spore_print_color','population', 'habitat']
# Displaying the first 5 rows
mushroom.head()
```

```python
print('Data shape:{}'.format(mushroom.shape))
print('Number of missing values:{}'.format(mushroom.isnull()))
```

```python
# Data organization into four explanatory variables
# Using dummy encoding to change all the variables into numerical values

mushroom_dummy = pd.get_dummies(mushroom[['gill_color','gill_attachment','odor','cap_color']], dtype = 'int')
mushroom_dummy.head()

# if its gill_colour 1 to represent black and 0 to represent the other

#Also converting the target variable since it is categorrical
mushroom_dummy['flg'] = mushroom['classes'].map(lambda x: 1 if x == 'p' else 0)


```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

#data split
X=mushroom_dummy.drop('flg',axis =1)
Y=mushroom_dummy['flg']
x_train, x_test , y_train , y_test = train_test_split(X ,Y ,test_size=0.2 ,random_state=0)

#intialize and train the decision tree classifier
model = DecisionTreeClassifier(criterion='entropy',max_depth=3,random_state=0)
model.fit(x_train,y_train)

print('Accuracy(train):{:.3f}'.format(model.score(x_train,y_train)))
print('Accuracy(test):{:.3f}'.format(model.score(x_test,y_test)))
```
